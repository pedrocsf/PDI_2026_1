## Atividade 1. Espectro e escalas

**Solicitado:** Calcular a FFT 2D de uma imagem e exibir a magnitude (a) bruta e (b) centralizada com `fftshift`. Comparar as escalas linear e logarítmica. Explicar por que o termo DC domina o espectro e qual o papel da escala log `s = log(1 + |F|)`.

**Observações:**

- A imagem de teste é um quadrado branco de 40x40 pixels (20 pixels para cada lado do centro) sobre fundo preto, em uma grade de 1024x1024.
- Na **magnitude bruta** (sem `fftshift`), o componente DC fica nos quatro cantos da imagem, pois a frequência zero está no índice (0,0). O valor máximo medido foi 1,60e3 e o mínimo 0, resultando em dinâmica da ordem de 1,6e13 vezes.
- Após o **`fftshift`**, o componente DC vai para o centro (512,512), e o espectro passa a ser lido de forma intuitiva: baixas frequências no centro, altas frequências nas bordas.
- **Por que o DC domina:** o termo DC corresponde à soma de todas as intensidades da imagem. Como o quadrado possui 40x40 = 1600 pixels de valor 1, o pico DC vale exatamente 1600, enquanto as demais componentes são ordens de grandeza menores. Por isso, na escala linear, apenas o ponto central aparece brilhante e todo o restante fica praticamente preto.
- **Papel da escala log:** a escala `s = log(1 + |F|)` comprime a faixa dinâmica, amplificando os valores pequenos e atenuando o pico. Isso revela a estrutura de lóbulos (padrão sinc) que a escala linear esconde.
- **Confirmação numérica:** a dinâmica do perfil horizontal foi de cerca de 1,6e13 vezes; a diferença máxima entre a escala linear e a log de Gonzalez chegou a 0,5941 e a correlação entre elas foi de 0,6995. Ou seja, mesmo após a normalização as escalas não são equivalentes quando a dinâmica é alta.

**Observações:**

- O espectro do quadrado é o padrão **sinc 2D**, com lóbulos alinhados em cruz nas direções x e y, coerente com a função sinc citada na teoria (transformada de uma janela retangular).

- **Borda nítida** (quadrado e círculo) gera altas frequências fortes; uma **transição suave** geraria altas frequências fracas, deixando o espectro mais concentrado no centro.
- **Sobre girar o quadrado:** girar a forma no espaço gira o espectro do mesmo ângulo. A estrutura de lóbulos acompanha a orientação do objeto, sem mudar o conteúdo de magnitude.

## Atividade 3. Propriedades

**Solicitado:** Verificar experimentalmente que a magnitude do espectro é invariante à translação do objeto (mudar a posição: a magnitude não muda, a fase sim). Comentar a simetria do espectro de uma imagem real e a separabilidade da FFT 2D.

**Observações:**

- **Invariância à translação:** comparando o quadrado no centro e no canto, a magnitude do espectro é idêntica. Apenas a fase muda com a posição (a fase não é visualizada no caderno). Isso confirma a propriedade `F[f(x-x0, y-y0)] = F(u,v) * e^(-j 2pi (u x0 + v y0))`, em que o fator exponencial afeta somente a fase, deixando a magnitude inalterada.
- **Simetria:** como a imagem de entrada é real, o espectro de magnitude apresenta simetria conjugada (hermitiana) em relação ao centro. Por isso os lóbulos aparecem espelhados em torno do ponto central.
- **Separabilidade:** a FFT 2D equivale a aplicar a FFT 1D nas linhas e, em seguida, a FFT 1D nas colunas. No espectro do quadrado isso se manifesta como o produto de dois sincs, formando a cruz alinhada aos eixos.

**Observações:**

- A **linha horizontal** concentra energia ao longo do eixo vertical do espectro, e a **linha vertical** concentra energia ao longo do eixo horizontal. Ou seja, a orientação da concentração de energia no espectro é perpendicular à orientação da estrutura no espaço.
- Por extensão da teoria e da tabela do resumo didático: um padrão periódico (listras) produz **picos discretos em frequências fixas**. Quanto mais fino e apertado o padrão (maior frequência espacial), mais **afastados do centro** ficam os picos. A direção da reta que liga os picos ao centro indica a **orientação** do padrão.
